checking for outlier (iqr)

```python
numerical_columns = [
    "rate (out of 5)",
    "num of ratings",
    "avg cost (two people)"
]
```

```python
for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    minimum = df[column].min()
    maximum = df[column].max()

    print("\nColumn:", column)
    print("Minimum:", minimum)
    print("Maximum:", maximum)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)

    if minimum < lower_limit or maximum > upper_limit:

        print("Outliers are present.")

        outliers = df[
            (df[column] < lower_limit) |
            (df[column] > upper_limit)
        ]

        print("Number of outliers:", len(outliers))

    else:
        print("No outliers outside the IQR limits.")
```

```python
print(df["num of ratings"].value_counts().sort_index())
```

removing outliers other than no of rating


```python
columns_to_remove_outliers = [
    "rate (out of 5)",
    "avg cost (two people)"
]

for column in columns_to_remove_outliers:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df = df[
        (df[column] >= lower_limit) &
        (df[column] <= upper_limit)
    ]

print("Shape after removing outliers:", df.shape)
```

checking for some suspicious characters from restaurant name


```python
characters = set()
for name in df["restaurant name"]:
    characters.update(name)
    
print(sorted(characters))
```

defining a text column including columns from data set

```python
text_columns = [
    "restaurant name",
    "restaurant type",
    "online_order",
    "table booking",
    "cuisines type",
    "area",
    "local address"
]
```

Check all text columns for non-ASCII characters

```python
for column in text_columns:
    suspicious = df[
        df[column].astype(str).apply(
            lambda x: any(ord(char) > 127 for char in x)
        )
    ]

    print("\n", "=" * 50)
    print("Column:", column)
    print("Suspicious rows:", len(suspicious))

    if len(suspicious) > 0:
        print(suspicious[[column]].drop_duplicates().to_string(index=False))
```

found that there is suspicious rows only in restaurant name

```python
def fix_text(name):
    
    fixed = name
    for _ in range(10):
        new_fixed = codecs.decode(fixed.encode("latin1"), "utf-8", "ignore")
        if new_fixed == fixed:
            break
        fixed = new_fixed
    return fixed

df["restaurant name"] = df["restaurant name"].apply(fix_text)
```
