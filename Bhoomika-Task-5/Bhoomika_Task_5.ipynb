```python
import pandas as pd
df = pd.read_csv("zomato.csv")
print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
print(df["Unnamed: 0.1"].equals(df["Unnamed: 0"]))
print(df["Unnamed: 0"].equals(pd.Series(range(len(df)))))
```

```python
df = df.drop(["Unnamed: 0.1", "Unnamed: 0"], axis=1)

print(df.head())
print(df.shape)
```

```python
print("Median rate:", df["rate (out of 5)"].median())
print("Median cost:", df["avg cost (two people)"].median())
```

```python
df["rate (out of 5)"] = df["rate (out of 5)"].fillna(df["rate (out of 5)"].median())

df["avg cost (two people)"] = df["avg cost (two people)"].fillna(df["avg cost (two people)"].median())
```
